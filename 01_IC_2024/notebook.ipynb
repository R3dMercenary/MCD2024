## Which decade had the highest ratio of US-born Nobel Prize winners to total winners in all categories?

```python
US_winners=df[df['birth_country']==top_country]
year=US_winners.groupby('year').agg({'birth_country':'value_counts'})
year
```

```python
decades=[1900,1910,1920,1930,1940,1950,1960,1970,1980,1990,2000,2010,2020]
decade_dictionary={}
for decade in decades:
    decade_dictionary[decade]=0
    
for year_country in year.index:
    year_value=year_country[0]
    number_of_winners=year.loc[(year_value,top_country),'birth_country']
    for decade in decades:
        in_decade= ( (year_value>decade) & (year_value<=decade+10) )
        if in_decade:
            decade_dictionary[decade]+=number_of_winners
            break
        else:
            continue
    
   
        
decade_dictionary
```
